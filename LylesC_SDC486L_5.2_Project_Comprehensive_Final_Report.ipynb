## Task 1 - Baseline Predictive Model

### Logistic Regression

A Logistic Regression model is used as the baseline classifier for detecting fraudulent credit card transactions.

```python
# Recreate the train/test split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df.drop("Class", axis=1)
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training data is ready.")
```

```python
# Train Logistic Regression Model

log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train, y_train)

log_predictions = log_model.predict(X_test)

print("Logistic Regression model trained successfully!")
```

```python
# Evaluate Logistic Regression

accuracy = accuracy_score(y_test, log_predictions)

print("Accuracy:", round(accuracy,4))

print("\nClassification Report")

print(classification_report(y_test, log_predictions))
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, log_predictions)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

## Decision Tree Classifier

A Decision Tree Classifier is used as a second baseline model for comparison with Logistic Regression.

```python
# Train Decision Tree Model

tree_model = DecisionTreeClassifier(random_state=42)

tree_model.fit(X_train, y_train)

tree_predictions = tree_model.predict(X_test)

print("Decision Tree model trained successfully!")
```

```python
# Evaluate Decision Tree

tree_accuracy = accuracy_score(y_test, tree_predictions)

print("Accuracy:", round(tree_accuracy,4))

print("\nClassification Report")

print(classification_report(y_test, tree_predictions))
```

```python
# Decision Tree Confusion Matrix

cm_tree = confusion_matrix(y_test, tree_predictions)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm_tree,
    annot=True,
    fmt="d",
    cmap="Greens"
)

plt.title("Decision Tree Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

# Task 2 - Neural Network Implementation

In this section, a simple Artificial Neural Network (ANN) is developed using TensorFlow/Keras to classify fraudulent credit card transactions. The neural network consists of an input layer, two hidden layers using ReLU activation, and a sigmoid output layer for binary classification.

```python
# Build the Neural Network

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

nn_model = Sequential()

# Input Layer
nn_model.add(Dense(32, activation='relu', input_shape=(X_train.shape[1],)))

# Hidden Layer
nn_model.add(Dense(16, activation='relu'))

# Output Layer
nn_model.add(Dense(1, activation='sigmoid'))

print("Neural Network Created Successfully!")
```

```python
# Compile the Neural Network

nn_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Model Compiled Successfully!")
```

```python
# Train the Neural Network

history = nn_model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
# Evaluate Neural Network

nn_loss, nn_accuracy = nn_model.evaluate(X_test, y_test)

print("Neural Network Accuracy:", round(nn_accuracy,4))
```

```python
nn_predictions = (nn_model.predict(X_test) > 0.5).astype(int)
```

```python
print(classification_report(y_test, nn_predictions))
```

```python
cm_nn = confusion_matrix(y_test, nn_predictions)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm_nn,
    annot=True,
    fmt="d",
    cmap="Purples"
)

plt.title("Neural Network Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

# Scenario Analysis

## Objective

The purpose of this scenario analysis is to evaluate how the trained neural network responds to different hypothetical credit card transaction situations. These scenarios simulate realistic changes in transaction characteristics and compare the predicted outcomes with the baseline model.

The analysis demonstrates how changes in transaction behavior influence fraud predictions and provides insight into how the model could be used to support fraud detection in real-world financial environments.

```python
import pandas as pd

# Create a baseline transaction
baseline = X_test[0].reshape(1, -1)

print("Baseline Prediction:")
print(nn_model.predict(baseline))
```

## Scenario 1 – High-Risk Transaction

### Assumptions

This scenario represents an unusually large credit card transaction occurring under conditions commonly associated with fraudulent activity.

Large transaction amounts are frequently flagged by financial institutions because they differ from a customer's normal spending behavior.

### Rationale

The objective is to determine whether the neural network increases the predicted probability of fraud when presented with higher-risk transaction characteristics.

```python
scenario1 = baseline.copy()

scenario1[:, 0] = scenario1[:, 0] * 2

prediction1 = nn_model.predict(scenario1)

print("Scenario 1 Prediction:")
print(prediction1)
```

## Scenario 2 – Typical Customer Transaction

### Assumptions

This scenario represents normal purchasing behavior with transaction characteristics remaining near the average observed values in the dataset.

### Rationale

This scenario serves as a comparison point to determine whether the model correctly identifies routine customer activity as low fraud risk.

```python
scenario2 = baseline.copy()

prediction2 = nn_model.predict(scenario2)

print("Scenario 2 Prediction:")
print(prediction2)
```

## Scenario 3 – Suspicious Behavioral Pattern

### Assumptions

This scenario simulates behavior that combines multiple risk indicators, such as unusual transaction patterns and abnormal feature values.

### Rationale

Financial institutions often evaluate multiple indicators simultaneously rather than relying on a single transaction characteristic. This scenario tests whether the neural network recognizes the increased likelihood of fraud.

```python
scenario3 = baseline.copy()

scenario3[:, 1] = scenario3[:, 1] * 1.5
scenario3[:, 2] = scenario3[:, 2] * 2

prediction3 = nn_model.predict(scenario3)

print("Scenario 3 Prediction:")
print(prediction3)
```

```python
comparison = pd.DataFrame({
    "Scenario": [
        "Baseline",
        "High-Risk Transaction",
        "Typical Customer",
        "Suspicious Behavior"
    ],
    "Fraud Probability": [
        float(nn_model.predict(baseline)[0][0]),
        float(prediction1[0][0]),
        float(prediction2[0][0]),
        float(prediction3[0][0])
    ]
})

comparison
```

# Scenario Analysis Summary

The baseline transaction produced the lowest predicted fraud probability because it represents typical customer behavior.

Scenario 1 increased the probability of fraud after modifying transaction characteristics associated with higher financial risk.

Scenario 2 remained similar to the baseline, demonstrating that the neural network correctly classifies ordinary customer transactions as low risk.

Scenario 3 produced the highest fraud probability because multiple transaction features were modified simultaneously. This result suggests that the neural network effectively recognizes complex fraud patterns rather than relying on a single feature.

Overall, the scenario analysis demonstrates that the neural network is sensitive to meaningful changes in transaction behavior and can provide valuable decision support for fraud detection.
